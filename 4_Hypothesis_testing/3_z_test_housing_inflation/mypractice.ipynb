```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("house_price_increase.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
population_mean=10
population_std_dev=4
```

```python
sample_mean=df.price_increase_pct.mean()
sample_mean
```

```python
std_error=population_std_dev/np.sqrt(df.shape[0])
std_error
```

```python
z_score=(sample_mean - population_mean)/std_error
z_score
```

```python
alpha=0.05 # significance level we took 0.05 because it is famous as 5% significance value and 95% confidence
```

```python
import scipy.stats as st
```

```python
z_critical=st.norm.ppf(1-alpha)
z_critical
```

```python
z_score, z_critical
```
