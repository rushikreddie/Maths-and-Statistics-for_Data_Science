#### (2) Adidas Analysis

```python
df_adidas=df[df.brand=="Adidas"]
df_adidas.shape
```

```python
df_adidas.describe()
```

```python
round(df_adidas.sold_qty.median())
```

```python
df_adidas.sold_qty.quantile(0.95)
```

```python
df_adidas[df_adidas.sold_qty>18.54]
```

```python
val=round(df_adidas.sold_qty.median())
```

```python
val
```

```python
df_adidas.sold_qty.replace(689,val,inplace=True)
```

```python
dates=df_nike.date
# Plot
plt.figure(figsize=(15,6))

plt.plot(dates,df_nike["sold_qty"],label="Nike",marker="o",color="blue")
plt.plot(dates,df_adidas["sold_qty"],label="Adidas",marker="o",color="red")
plt.title('Nike vs Adidas Sold Quantity')
plt.xlabel('Date')
plt.ylabel('Sold Quantity')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```
