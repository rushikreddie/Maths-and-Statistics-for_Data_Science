```python
import numpy as np
```

```python
expected=np.array([40,60,30,70])
actual=np.array([45,55,35,65])
```

```python
import scipy.stats as st
```

```python
chi_squared_statistic,p_value=st.chisquare(actual,expected)
```

```python
chi_squared_statistic,p_value
```

```python
significance_value=0.05
```

```python
p_value < significance_value
```
