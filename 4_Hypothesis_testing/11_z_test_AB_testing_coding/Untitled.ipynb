```python
import pandas as pd
import numpy as np
from scipy import stats as st
```

```python
control=pd.read_csv("control_group.csv")
control.head()
```

```python
test=pd.read_csv("test_group.csv")
```

```python
test.head()
```

```python
control_mean=control.recovery_time_hrs.mean().round(2)
control_std_dev=control.recovery_time_hrs.std().round(2)
control_size=control.shape[0]
control_mean, control_std_dev, control_size
```

```python
test_mean=test.recovery_time_hrs.mean().round(2)
test_std_dev=test.recovery_time_hrs.std().round(2)
test_size=test.shape[0]
test_mean, test_std_dev, test_size
```

```python
a=control_std_dev**2/control_size
b=test_std_dev**2/test_size
z_score=(control_mean-test_mean)/np.sqrt(a+b)
z_score
```

```python
Z_critical=st.norm.ppf(1-0.05)
Z_critical
```

```python
z_score>Z_critical
```

```python
p_value=1-st.norm.cdf(z_score)
p_value
```

```python
p_value<0.05
```
