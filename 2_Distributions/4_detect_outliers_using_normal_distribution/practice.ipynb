```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("heights_with_outliers.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
mean=round(df.height.mean(),2)
std=round(df.height.std(),2)
mean, std 
```

```python
sns.histplot(df.height,kde=True)
# Draw vertical lines
plt.axvline(mean, color='black', label='Mean')
plt.axvline(mean + std, color='green', linestyle='--', label='Mean + 1σ')
plt.axvline(mean - std, color='green', linestyle='--', label='Mean - 1σ')

plt.axvline(mean + 2*std, color='red',linestyle='--',label='Mean + 2σ')
plt.axvline(mean - 2*std, color='red',linestyle='--',label='Mean + 2σ')

plt.axvline(mean + 3*std, color='red',linestyle='--',label='Mean + 3σ')
plt.axvline(mean - 3*std, color='red',linestyle='--',label='Mean + 3σ')

plt.legend()
plt.title("Height Distribution with Mean ± 1σ")

plt.show()
```

```python
lower=mean-3*std
upper=mean+3*std
lower,upper
```

```python
df[(df["height"]<lower) | (df["height"] > upper)]
```

```python
df_new=df[(df["height"]>lower) & (df["height"] < upper)]
df_new.shape
```

```python
sns.histplot(df_new.height,kde=True)
plt.show()
```
