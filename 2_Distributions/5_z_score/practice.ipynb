```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("heights_with_outliers.csv")
```

```python
df.head()
```

```python
mean=df.height.mean()

std=df.height.std()
df['Zscore']=(df['height']-mean)/std
```

```python
df.head()
```

```python
df[(df.Zscore < -3) | (df.Zscore > 3)]
```

```python
df_new=df[(df.Zscore > -3) & (df.Zscore < 3)]
```

```python
df_new.shape
```

```python
sns.histplot(df_new.height)
```

```python
sns.histplot(df_new.Zscore)
```
