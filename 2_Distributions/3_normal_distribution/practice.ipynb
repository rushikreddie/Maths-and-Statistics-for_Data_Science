```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns 
```

```python
df=pd.read_csv("heights.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
sns.histplot(df.height,kde=True)
mean = df['height'].mean()
std = df['height'].std()


# Draw vertical lines
plt.axvline(mean, color='black', label='Mean')
plt.axvline(mean + std, color='green', linestyle='--', label='Mean + 1σ')
plt.axvline(mean - std, color='green', linestyle='--', label='Mean - 1σ')

plt.axvline(mean + 2*std, color='red',linestyle='--',label='Mean + 2σ')
plt.axvline(mean - 2*std, color='red',linestyle='--',label='Mean + 2σ')

plt.axvline(mean + 3*std, color='red',linestyle='--',label='Mean + 3σ')
plt.axvline(mean - 3*std, color='red',linestyle='--',label='Mean + 3σ')

plt.legend()
plt.title("Height Distribution with Mean ± 1σ")

plt.show()
```

```python
left1=round(mean-std,2)
right1=round(mean+std,2)
left1,right1
```

```python
left2=round(mean-2*std,2)
right2=round(mean+2*std,2)
left2,right2
```

```python
left3=round(mean-3*std,2)
right3=round(mean+3*std,2)
left3,right3
```
